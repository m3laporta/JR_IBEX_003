```python
# Cell 5: Set Sample Context and Data Paths
print("📂 Setting up data context...")

# Check for existing data
data_paths = {
    'raw': project_root / "data" / "raw",
    'processed': project_root / "data" / "processed", 
    'aligned': project_root / "aligned",
    'results': project_root / "results"
}

# Check data availability
print("📁 Data directories:")
for name, path in data_paths.items():
    if path.exists():
        files = list(path.glob("*"))
        print(f"   ✅ {name}: {path} ({len(files)} items)")
    else:
        print(f"   ⚠ {name}: {path} - NOT FOUND")

# Look for existing samples
sample_dirs = []
aligned_dir = project_root / "aligned"
if aligned_dir.exists():
    sample_dirs = [d.name for d in aligned_dir.iterdir() if d.is_dir()]

if sample_dirs:
    print(f"\n🎯 Available samples: {sample_dirs}")
    
    # Check if we have a current sample from session state
    try:
        state_path = project_root / "config" / "daily_state.json"
        if state_path.exists():
            with open(state_path, 'r') as f:
                state = json.load(f)
            current_sample = state.get('current_sample', {}).get('roi_id')
            if current_sample:
                print(f"   📌 Current sample from session: {current_sample}")
            else:
                print("   📌 No current sample set - you can set one in the next cell")
    except:
        pass
else:
    print("\n📝 No existing samples found. You can:")
    print("   1. Place raw data in data/raw/")
    print("   2. Run alignment workflow to create aligned samples")
    print("   3. Or continue with method setup and testing")

# Show existing notebooks integration
notebooks_dir = project_root / "notebooks"
if notebooks_dir.exists():
    existing_notebooks = list(notebooks_dir.glob("*.py"))
    print(f"\n🔧 Existing IBEX notebooks: {[nb.name for nb in existing_notebooks]}")
    print("   These are fully integrated and available via existing workflow")
```

```python
# Cell 6: Create Helper Functions and Shortcuts
print("🛠️ Setting up helper functions...")

def set_current_sample(roi_id):
    """Set the current sample for analysis"""
    try:
        from scripts.resume_session import load_daily_state, save_daily_state
        state = load_daily_state()
        state['current_sample']['roi_id'] = roi_id
        save_daily_state(state)
        print(f"✅ Current sample set to: {roi_id}")
        return roi_id
    except Exception as e:
        print(f"⚠ Failed to set sample: {e}")
        return None

def load_sample_data(roi_id=None):
    """Load data for the current or specified sample"""
    if roi_id is None:
        try:
            from scripts.resume_session import load_daily_state
            state = load_daily_state()
            roi_id = state.get('current_sample', {}).get('roi_id')
        except:
            pass
    
    if not roi_id:
        print("❌ No sample ID provided. Use set_current_sample() first")
        return None
    
    # Look for aligned data
    aligned_path = project_root / "aligned" / roi_id
    if aligned_path.exists():
        aligned_files = list(aligned_path.glob("*.tif"))
        print(f"📊 Found {len(aligned_files)} aligned files for {roi_id}")
        return {'roi_id': roi_id, 'aligned_path': aligned_path, 'files': aligned_files}
    else:
        print(f"⚠ No aligned data found for {roi_id}")
        return None

def quick_analysis(roi_id=None):
    """Run quick spatial analysis on current sample"""
    sample_data = load_sample_data(roi_id)
    if sample_data:
        print(f"🔬 Running quick analysis on {sample_data['roi_id']}")
        # Placeholder for actual analysis
        print("   📊 This would run comprehensive spatial analysis")
        print("   📊 Using all available literature methods")
        return sample_data
    return None

def show_progress():
    """Show current analysis progress"""
    try:
        summary = progress_tracker.get_session_summary()
        print("📊 Session Progress:")
        for key, value in summary.items():
            print(f"   {key}: {value}")
    except:
        print("⚠ Progress tracking not available")

def save_checkpoint(name, description=""):
    """Create a progress checkpoint"""
    try:
        from scripts.track_progress import checkpoint
        checkpoint_path = checkpoint(name, description)
        print(f"✅ Checkpoint '{name}' saved")
        return checkpoint_path
    except Exception as e:
        print(f"⚠ Checkpoint failed: {e}")

# Make functions globally available
globals().update({
    'set_current_sample': set_current_sample,
    'load_sample_data': load_sample_data,
    'quick_analysis': quick_analysis,
    'show_progress': show_progress,
    'save_checkpoint': save_checkpoint
})

print("✅ Helper functions ready:")
print("   - set_current_sample(roi_id)")
print("   - load_sample_data(roi_id=None)")
print("   - quick_analysis(roi_id=None)")
print("   - show_progress()")
print("   - save_checkpoint(name, description)")

# Update progress
try:
    update_progress('00_daily_startup', 'running', 75)
except:
    pass
```

```python
# Cell 7: Final Setup and Daily Goals
print("🎯 Final setup and daily planning...")

# Show session summary
print("\n📋 Today's Session Summary:")
print("=" * 40)

try:
    from scripts.resume_session import load_daily_state
    state = load_daily_state()
    
    # Current sample
    current_sample = state.get('current_sample', {}).get('roi_id')
    if current_sample:
        print(f"📊 Current sample: {current_sample}")
    else:
        print("📊 No current sample set")
    
    # Daily goals
    goals = state.get('notes', {}).get('daily_goals', [])
    if goals:
        print(f"\n🎯 Daily goals ({len(goals)}):")
        for i, goal in enumerate(goals, 1):
            print(f"   {i}. {goal}")
    else:
        print("\n🎯 No daily goals set - consider adding some!")
    
    # Completed tasks
    completed = state.get('notes', {}).get('completed_tasks', [])
    if completed:
        recent_completed = completed[-3:]  # Show 3 most recent
        print(f"\n✅ Recent completions:")
        for task in recent_completed:
            print(f"   ✓ {task}")

except Exception as e:
    print(f"⚠ Could not load session state: {e}")

# Show available methods summary
print(f"\n🔬 Available Analysis Methods:")
if literature_methods:
    for method_name in literature_methods.keys():
        print(f"   ✅ {method_name} lab methods")
        
print("\n📚 Available Workflows:")
workflow_files = list((project_root / "workflows").glob("*.ipynb")) if (project_root / "workflows").exists() else []
for workflow in sorted(workflow_files):
    print(f"   📖 {workflow.name}")

# Suggest next steps
print("\n🚀 Suggested Next Steps:")
print("=" * 30)
if not current_sample:
    if sample_dirs:
        print(f"1. Set current sample: set_current_sample('{sample_dirs[0]}')")
    else:
        print("1. Place raw data in data/raw/ and run alignment")
    print("2. Or explore methods with test data")
else:
    print(f"1. Continue analysis of {current_sample}")
    print("2. Run: quick_analysis() for comprehensive spatial analysis")
    print("3. Or proceed to specific workflow notebooks")

print("\n💡 Pro Tips:")
print("   - Use save_checkpoint('name') to save progress anytime")
   "   - All your work is automatically tracked and saved")
print("   - Literature methods are ready to use immediately")
print("   - Check show_progress() to see your session stats")

# Complete startup
try:
    update_progress('00_daily_startup', 'completed', 100)
    print("\n✅ Daily startup completed successfully!")
except:
    print("\n✅ Daily startup completed!")

print("\n" + "="*60)
print("🎉 JR_IBEX_003 MASTER ORGANIZATION SYSTEM READY! 🎉")
print("   You can now continue exactly where you left off!")
print("="*60)
```
